```python
import pandas as pd
import numpy as np
```

```python
telco_data = pd.read_csv('Telco-Customer-Churn.csv')
```

```python
df = telco_data.copy()
```

```python
import inflection
df.columns = [
    inflection.underscore(col)
    for col in df.columns
]
```

```python
df.columns
```

```python
df.nunique()
```

```python
service_columns = ['multiple_lines','online_security', 'online_backup', 'device_protection', 'tech_support','streaming_tv', 'streaming_movies']
for col in service_columns:
    print(df[col].value_counts())
```

```python
for col in service_columns:
    df[col] = df[col].replace({'No phone service':'No','No internet service':'No'})
```

```python
df.nunique()
```

```python
df.isna().sum()
```

```python
df['total_charges'] = pd.to_numeric(df['total_charges'],errors='coerce')
```

```python
df.info()
```

```python
df['senior_citizen'] = df['senior_citizen'].map({0:'No',1:'Yes'})
```

```python
df.head(2)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df[df['total_charges'].isna()]
```

```python
ind1 = df[df['total_charges'].isna()].index
df.loc[ind1,'total_charges'] = 0
```

```python
df.isna().sum()
```

```python
df.to_csv('telco-cleaned-data.csv',index=False)
```
